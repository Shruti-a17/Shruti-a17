```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load iris dataset
iris = load_iris()
X_all = iris.data
Y_all = iris.target

# Keep only Setosa (0) and Versicolor (1)
mask = (Y_all == 0) | (Y_all == 1)
X1 = X_all[mask, 0]   # sepal length
X2 = X_all[mask, 2]   # petal length
Y  = Y_all[mask]      # labels (0 or 1)

n = len(Y)

# Initialize weights
w0, w1, w2 = 0.0, 0.0, 0.0
alpha = 0.01
epochs = 1000

# Gradient descent
for epoch in range(epochs):
    # Predictions
    Z = w0 + w1*X1 + w2*X2   # z_i for all samples

    # Errors
    error = Y - Z            # y_i - z_i

    # Gradients (mathematical formulas)
    dE_dw0 = -(1/n) * np.sum(error)
    dE_dw1 = -(1/n) * np.sum(error * X1)
    dE_dw2 = -(1/n) * np.sum(error * X2)

    # Update weights
    w0 = w0 - alpha * dE_dw0
    w1 = w1 - alpha * dE_dw1
    w2 = w2 - alpha * dE_dw2

    # Print error every 100 epochs
    if epoch % 100 == 0:
        E = (1/(2*n)) * np.sum(error**2)
        print(f"Epoch {epoch}, Error={E:.4f}")

print("Final weights:", w0, w1, w2)
```
